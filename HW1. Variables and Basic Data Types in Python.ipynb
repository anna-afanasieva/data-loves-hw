```python
# 2. Арифметичні операції

# Створіть дві змінні: var1 зі значенням 15, var2 зі значенням 4.

# Обчисліть та виведіть (кожен пункт оцінюється в 0.5 бала):

# Суму
# Різницю
# Добуток
# Ділення
# Цілочисельне ділення
# Залишок від ділення
# Піднесення до степеня
```

```python
# Arithmetic Operations

var_1 = 15
var_2 = 4

print(var_1 + var_2)   
print(var_1 - var_2)   
print(var_1 * var_2)   
print(var_1 / var_2)  
print(var_1 // var_2)  
print(var_1 % var_2)   
print(var_1 ** var_2)
```
