5. Статистика із групуванням

Обчисліть (кожен пункт оцінюється в 0.5 бала):

- Середній рахунок за статтю (groupby sex)
- Середні чайові за днем тижня (groupby day)
- Скільки відвідувачів у кожен день

```python
# Групування за однією колонкою
#Середній рахунок за статтю (groupby sex)

by_sex = df.groupby('sex')['total_bill'].mean()
print("Середній рахунок за статтю:")
by_sex
```

```python
by_sex = df.groupby('sex')['total_bill'].mean().reset_index()
display(by_sex)
```

```python
#Середні чайові за днем тижня (groupby day)

by_day = df.groupby('day')['tip'].mean().reset_index().round(2)
display(by_day)
```

```python
#Скільки відвідувачів у кожен день
by_day = df.groupby('day')['day'].count().reset_index(name='count')
display(by_day)
```

6. Створення нової колонки

Створіть нову колонку tip_percentage - відсоток чайових від загальної суми рахунку. Виведіть перші 10 рядків з новою колонкою.

```python
df['tip_percentage'] = ((df['tip'] / df['total_bill']) * 100).round(2)
display(df['tip_percentage'].head(10))
#or
display(df[['total_bill', 'tip', 'tip_percentage']].head(10))
```
