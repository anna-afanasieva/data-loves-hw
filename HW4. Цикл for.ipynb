```python
# 2. Цикл for по списку

# Дано список:


# [ ]
# names = ["Олег", "Анна", "Іван", "Марія"]
# Використовуючи цикл, виведіть кожне ім'я з привітанням: "Привіт, [ім'я]!"
```

```python
# 3. Цикл з enumerate

# Дано список оцінок:


# [ ]
# grades = [85, 92, 78, 95, 88]
# Використовуючи enumerate, виведіть кожну оцінку з номером: "Студент 1: 85 балів"
```

```python
# enumerate — це функція яка до кожного елемента списку додає його порядковий номер.
# i — ловить номер (0, 1, 2, 3, 4)
# grade — ловить оцінку (85, 92, 78, 95, 88)

grades = [85, 92, 78, 95, 88]

for i, grade in enumerate(grades):
    print(f"Студент {i+1}: {grade} балів")
```

```python
# enumerate(grades) - так Python бачить список

grades = [85, 92, 78, 95, 88]

for i in enumerate(grades):
    print(i)
```

```python
# f"Студент {i+1}, бо буде 0 першим

grades = [85, 92, 78, 95, 88]

for i, grade in enumerate(grades):
    print(f"Студент {i}: {grade} балів")
```

```python
# 4. Сума парних чисел

# Напишіть програму, яка обчислює суму всіх парних чисел від 1 до 20 включно.
```

```python
for i in range(1, 21):
    print(i)
```

```python
total = 0
for i in range(1, 21):
    if i % 2 == 0:
        total += i
print(total)
```
