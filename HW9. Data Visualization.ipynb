Про датасет:

Tips - це датасет, який містить інформацію про рахунки та чайові в ресторані.

Колонки датасету:

- total_bill - загальна сума рахунку (долари)
- tip - розмір чайових (долари)
- sex - стать відвідувача (Male/Female)
- smoker - чи курить (Yes/No)
- day - день тижня (Thur/Fri/Sat/Sun)
- time - час відвідування (Lunch/Dinner)
- size - розмір групи (кількість людей)
Джерело: Оригінальні дані зібрані в одному з американських ресторанів у 1990-х роках. Скачати і переглянути набір даних можна тут: https://www.kaggle.com/datasets/sakshisatre/tips-dataset

1. Гістограма розподілу рахунків Завантажте датасет Tips (Чайові) з файла у матеріалах tips.csv. Створіть гістограму для колонки total_bill з 20 bins. Опціонально можна додати підписи осей ОХ, ОУ, а також назву графіку.

```python
import matplotlib.pyplot as plt
import pandas as pd


df = pd.read_csv('tip.csv')
```

```python
plt.hist(df['total_bill'], bins=20)
plt.show()
```

```python
df['total_bill'].hist(bins=20, edgecolor='red')
plt.title('Розподіл рахунків')
plt.xlabel('Сума рахунку ($)')
plt.ylabel('Кількість')
plt.show()
```

2. Box plot чайових за днями тижня

Створіть box plot, який показує розподіл чайових (tip) за днями тижня (day). В який день навище медіанне (середнє) значення чайоивх? Зелена горизонтальна лінія всередині кожного box-plot.

```python
# Boxplot 
df.boxplot(column='tip', by='day', figsize=(8, 5))
plt.title('Розподіл чайових за днями тижня')
plt.suptitle('')
plt.ylabel('Чайові')
plt.show()
```

```python
#В який день навище медіанне (середнє) значення чайоивх? 
#Відповідь: в неділю ($3.15)
```

3. Scatter plot: рахунок vs чайові

Створіть діаграму розсіювання для total_bill та tip.

```python
# Scatter plot - залежність суми рахунку і чайових
df.plot.scatter(x='total_bill', y='tip', alpha=0.5, c='size', colormap='viridis')
plt.title('Залежність між суммою рахунку і чайовими')
plt.xlabel('Рахунок')
plt.ylabel('Чайові')
plt.show()
```

4. Bar chart: середній рахунок за часом

Створіть стовпчасту діаграму середнього рахунку за часом дня (time).

```python
df.groupby('time')['total_bill'].mean().plot(kind='bar', color='skyblue', edgecolor='black')
plt.title('Середній рахунок за часом дня')
plt.xlabel('час відвідування')
plt.ylabel('загальна сума рахунку ($)')
plt.show()
```

5. Інтерактивна діаграма розсіювання

Створіть з допомогою бібліотек Plotly інтерактивний scatter plot для total_bill vs tip з кольором за статтю (sex).

```python
import plotly.express as px
```

```python
fig = px.scatter(df, x='total_bill', y='tip', color='sex',
                 hover_data=['day', 'time'],
                 title='Рахунок vs Чайові',
                 labels={'total_bill': 'Рахунок ($)', 'tip': 'Чайові ($)', 'sex': 'Стать'})
fig.show()
```

6. Інтерактивний box plot: чайові за днями та статтю

Створіть інтерактивний box plot для чайових за днями тижня з розбивкою за кольором по статі.

```python
fig = px.box(df, x='day', y='tip', color='sex',
             title='Чайові за днями тижня',
             labels={'day': 'День', 'tip': 'Чайові ($)', 'sex': 'Стать'})
fig.show()
```
