Завдання 1. Ми обговорили з Вами на лекції, що велика літера Сігма грецького алфавіту позначає в математиці суму. Знайдіть з використанням Python  Σ𝑥 ,  Σ𝑥2 , та  Σ(𝑥−1)2  для наступного набору даних: 1 3 4 Виведіть результат кожного обчислення на екран.

```python
x = [1, 3, 4]
```

```python
# Σx
# Просто сума всіх чисел 
# Σ (велика сигма) — означає суму
# σ (мала сигма) — означає стандартне відхилення в статистиці

print(sum(x))
```

```python
# Σ𝑥2
# Кожне число зводимо в квадрат, потім складаємо = 1² + 3² + 4² = 1 + 9 + 16 = 26

print(sum([i**2 for i in x]))
```

```python
# Σ(x-1)² 
# Від кожного числа віднімаємо 1, потім зводимо в квадрат, потім складаємо = (1-1)² + (3-1)² + (4-1)² = 0 + 4 + 9 = 13

print(sum([(i-1)**2 for i in x]))
```

Завдання 6. Крім даних про суму покупок, ми зібрали ще дані про кількість відвідувань кожного з кожного покупця, які знаходяться в змінній visits. Обчисліть коефіцієнт кореляції Пірсона для суми покупок і кількості відвідувань кожним з зазначених методів:

6.1. без використання жодних бібліотек, користуючись тільки Python.
    
6.2. використовуючи відповідний метод бібліотеки numpy.

Виведіть результат на екран. Проаналізуйте результат.

Виведіть додатково графік розсіювання викликавши код нижче і зробіть заключення, чи варто щось змінити при підрахунку кореляції, аби коефіцієнт Пірсона показував коректну оцінку лінійної залежності?

```python
sum_purchases = [ 316,  493,  367,  153,  224,  121,  322,  450,  196,  397,  281,
  239,  484,  124,  223,  448,  432,  255, 4310, 3524]

visits = [ 6,  2, 10,  3,  9,  9,  2,  3,  6,  6, 10,  2,  5, 10,  3,  5,  4,
        7,  4,  2]
```

```python
x = sum_purchases
y = visits
```

```python
#6.1. без використання жодних бібліотек, користуючись тільки Python.

mean_x = sum(x) / len(x)
mean_y = sum(y) / len(y)

mean_x
```

```python
mean_y
```

```python
cov_xy = sum([(x[i] - mean_x) * (y[i] - mean_y) for i in range(len(x))]) / (len(x) - 1)
print(cov_xy)
```

```python
std_x_sample = (sum([(el - mean_x)**2 for el in x]) / (len(x) - 1)) ** 0.5
std_y_sample = (sum([(el - mean_y)**2 for el in y]) / (len(y) - 1)) ** 0.5
print(std_x_sample, std_y_sample)
```

```python
corr = cov_xy / (std_x_sample * std_y_sample)
print(corr)
```

```python
# 6.2. використовуючи відповідний метод бібліотеки numpy.
np.corrcoef(x, y)
```

```python
# Виведіть додатково графік розсіювання викликавши код нижче і зробіть заключення, чи варто щось змінити при підрахунку кореляції, 
# аби коефіцієнт Пірсона показував коректну оцінку лінійної залежності?

import matplotlib.pyplot as plt

plt.scatter(x, y)
plt.xlabel('Сума покупок')
plt.ylabel('Кількість візитів')
plt.title('Залежність суми покупок та візитів')
plt.show()
```

Два викиди 3524 і 4310 спотворюють коефіцієнт Пірсона. Для коректної оцінки лінійної залежності напевно варто видалити ці викиди і перерахувати кореляцію.

Завдання 7. В попередньому завданні ми помітили, що викиди нам заважають отримати "правдиве" значення кореляції. Давайте вилучимо записи, які відповідають значенням з викидами в сумі покупок і обчисліть коеф. кореляції. Для цього поставте threshold, який дорінює середньому значенню + 2 стандартних вдіхилля суми покупок і всі значення, які більше за нього, будемо вважати викидами (є і інші правила для визначення викидів - з ним и познайомимось пізніше).

Вилучіть значення з обох масивів і обчисліть коеф. кореляції Пірсона з допомогою numpy для кількості візитів і суми покупок без записів-викидів. Виведіть результат на екран, побудуйте діаграму розсіювання і проаналізуйте результат: що можемо сказати про залежність в даних?

```python
threshold = mean_x + 2 * std_x_sample
print(threshold)
```

```python
x_clean = [x[i] for i in range(len(x)) if x[i] < threshold]
y_clean = [y[i] for i in range(len(x)) if x[i] < threshold]
print(x_clean)
print(y_clean)
```

```python
print(np.corrcoef(x_clean, y_clean))

plt.scatter(x_clean, y_clean)
plt.xlabel('Сума покупок')
plt.ylabel('Кількість візитів')
plt.title('Залежність без викидів')
plt.show()
```

Після видалення викидів кореляція змінилась з -0.30 до -0.37.
Залежність залишається слабкою від'ємною — чіткого лінійного зв'язку між сумою покупок і кількістю візитів немає.
