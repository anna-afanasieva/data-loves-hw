# Dictionaries: Basic Operations

```python
me = {
    "name": "Anna",
    "age": 30,
    "city": "Dnipro"   
}
```

```python
print(me["name"]) 
```

```python
print(me["age"])
```

```python
me["age"] = 21 
print(me["age"])
```

```python
me["hobby"] = "programming"
print(me["hobby"])
```

```python
print(me.keys())
```

```python
print(me.values()) 
```
