# 1. Basics of Conditional Statements (if-elif-else)

```python
cooking_time = 5

if cooking_time < 4:
    print("The egg is runny")
elif 4 <= cooking_time < 7:
    print("The egg is soft-boiled")
elif 7 <= cooking_time < 10:
    print("The egg is hard-boiled")
elif cooking_time >= 10:
    print ("The egg is overcooked")
```

# 2. Grade Check

```python
grade = 72

if grade >= 90:
    print("Excellent")
elif 75 <= grade < 90:
    print("Good")
elif 60 <= grade < 75:
    print("Satisfactory")
elif grade < 60:
    print("Unsatisfactory")
```

```python
products_in_the_fridge = ["milk", "eggs", "butter", "cheese"]
find_product = "eggs"
if find_product in products_in_the_fridge:
    print(f"There are {find_product} in the fridge")
else:
    print(f"There are no {find_product}")
```
