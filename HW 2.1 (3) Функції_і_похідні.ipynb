1. Задана функція залежності оцінки від кількості годин, витрачених на навчання: $ f(x) = 10x - 5 $.
  1. З допомогою Python обчисліть похідну цієї функції.
  2. Використовуйте бібліотеку `matplotlib` для побудови графіків функції та її похідної в діапазоні значень від 0 до 10 годин. Проаналізуйте, як зміна кількості годин вивчення впливає на оцінку.



```python
import numpy as np
import matplotlib.pyplot as plt
import sympy as sp
```

```python
x = sp.symbols('x')
```

```python
f = 10 * x - 5
```

```python
df = sp.diff(f, x)

df
```

```python
x = np.linspace(0, 10, 100)
f = 10 * x - 5
df = np.full_like(x, 10)
```

```python
plt.figure(figsize=(10, 5))

plt.plot(x, f, label='f(x) = 10x - 5', color='blue')
plt.plot(x, df, label="f'(x) = 10", color='green')

plt.xlabel('Hours of study')
plt.ylabel('Score')
plt.title('Function and derivative')
plt.legend()
plt.grid(True)
plt.show()
```

```python
Якщо х збільшується на 1 годину навчання, оцінка зростає на 10 балів.
```

2. Використовуючи Python і `SymPy`, обчисліть похідні для функцій $ f(x) = \sin(x) $ та $ f(x) = \ln(x) $. Зобразіть ці функції та їхні похідні за допомогою `matplotlib`.



```python
x = sp.symbols('x')
```

```python
f = sp.sin(x)
```

```python
df = sp.diff(f, x)

df
```

```python
x = np.linspace(-10, 10, 100)

f = np.sin(x)
df = np.cos(x)
```

```python
plt.figure(figsize=(10, 5))

plt.plot(x, f, label='f(x) = sin(x)', color='blue')
plt.plot(x, df, label="f'(x) = cos(x)", color='green')

plt.xlabel('x')
plt.ylabel('Value')
plt.title('Function sin(x) and its derivative')
plt.legend()
plt.grid(True)
plt.show()
```

```python
x1 = sp.symbols('x')
```

```python
f1 = sp.ln(x1)
```

```python
df1 = sp.diff(f1, x1)

df1
```

```python
x1_values = np.linspace(0.1, 10, 100)
```

```python
f1_values = np.log(x1_values)
df1_values = 1 / x1_values
```

```python
plt.figure(figsize=(10, 5))

plt.plot(x1_values, f1_values, label='f(x) = ln(x)', color='blue')
plt.plot(x1_values, df1_values, label="f'(x) = 1/x", color='green')

plt.xlabel('x')
plt.ylabel('Value')
plt.title('Function ln(x) and its derivative')
plt.legend()
plt.grid(True)
plt.show()
```


3. Загальні витрати компанії від обсягу виробництва задані функцією $ f(x) = x^2 - 5x + 1 $, де $ x $ — кількість вироблених одиниць. Обчисліть похідну цієї функції з допомогою SymPy та візуалізуйте значення функції і її похідної на відрізку від 0 до 100.

```python
x2 = sp.symbols('x')
```

```python
f2 = x2**2 - 5*x2 + 1
df2 = sp.diff(f2, x2)

df2
```

```python
x2_values = np.linspace(0, 100, 100)

f2_values = x2_values**2 - 5*x2_values + 1
df2_values = 2*x2_values - 5
```

```python
plt.figure(figsize=(10, 5))

plt.plot(x2_values, f2_values, label='f(x) = x² - 5x + 1', color='blue')
plt.plot(x2_values, df2_values, label="f'(x) = 2x - 5", color='green')

plt.xlabel('Number of produced units')
plt.ylabel('Value')
plt.title('Company costs function and its derivative')
plt.legend()
plt.grid(True)
plt.show()
```
